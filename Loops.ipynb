```python
## While loop
# continue to execute as long as condition is true
# if we will not increment, it will continue printing.
count=0
while count<5:
    print(count)
    count = count+1
```

```python
count=0
while count*2<10:
    print(count)
    count=count+1
```

```python
# Loop control statement

## break statement:  It is used to terminate the loop when a certain condition is met.
## cONTINUE statement:  It is used to skip the current iteration of the loop and move on to the next iteration.
## Pass statement:  It is used to skip the current iteration of the loop and move on to the next iteration.
for i in range(10):
    if i==5:
        break
    print(i)
```

```python
## cONTINUE statement:  It is used to skip the current iteration of the loop and move on to the next iteration.
for i in range(10):
    if i==5:
        continue
    print(i)
```

```python
## Pass statement:  It is used to skip the current iteration of the loop and move on to the next iteration.
for i in range(10):
    if i==5:
        pass
    print(i)
```

```python
# Nested Loop 
# a loop inside another loop is called nested loop. 
# The inner loop will be executed one time for each iteration of the outer loop.
for i in range(1,4):
    for j in range(1,4):
        print(i,j)

```

```python
for i in range(4):
    for j in range(2):
        print(i,j)

# when i is 0, j will be 0,1,2
# when i is 1, j will be 0,1,2
# when i is 2, j will be 0,1,2
# when i is 3, j will be 0,1,2
```

```python
# example : calculate the sum of first N natural number using a while and for loop

# while loop
n=20 # first 20 natural number
sum=0 # storing the sum in sum variable, initial is 0
count=1

while count<=n:
    sum=sum+count
    count=count+1
print("Sum of N natural number is :" , sum)
```

```python
## Display the prime number between 1 to 100

for num in range(1,101):
    if num>1:
        for i in range(2,num):
            if num%i==0: # if 
                break
        else:
            print(num)

        
```
