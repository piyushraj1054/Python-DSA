```python
## what is if statement in python?
# An "if statement" in Python is a control flow statement that allows you to execute a block of
# and code only if a specified condition evaluates to True. It is used to make decisions 
# in your code based on certain conditions. The basic syntax of an if statement is as follows:

age = 20
if age >= 18:
    print("You are an adult and eligible for voting.")

 
```

```python
## else statement in python?
# An "else statement" in Python is used in conjunction with an "if statement" to execute a 
# block of code when the condition in the if statement is False. The basic syntax of an if-else statement is as follows:

age = 15
if age >= 18:
    print("You are an adult and eligible for voting.")
else:
    print("You are not an adult and not eligible for voting.")
```

```python
# elif statement in python?
# An "elif statement" in Python is used to check multiple conditions in a sequence. 
# It  allows you to specify additional conditions to be checked if the previous conditions in the if-elif-else chain are False. 
# The basic syntax of an if-elif-else statement is as follows:  
age = 25
if age < 18:
    print("You are a minor and not eligible for voting.")
elif age >= 18 and age < 65:
    print("You are an adult and eligible for voting.")
else:
    print("You are a senior citizen and eligible for voting.")
```

```python
# Nested Conditional Statements in Python
# Nested conditional statements in Python refer to the practice of placing one or more conditional statements 
# (if, elif, else) inside another conditional statement. 
# This allows you to create more complex decision-making structures in your code.
num=int(input("Enter a number: "))
if num > 0:
    print("The number is positive.")
    if num % 2 == 0:
        print("The number is even.")
    else:
        print("The number is odd.")
else:
    print("The number is not positive.")
```

```python
## Practical example
# divisible by 100 but not by 400 is not a leap year
# divisible by 400 is a leap year
# divisible by 4 but not by 100 is a leap year

year = int(input("Enter a year: "))
if(year%4==0):
    if(year%100==0):
        if(year%400==0):
            print(year,"is a leap year")
        else:
            print(year," is not a leap year")
    else:
        print(year," is a leap year")    
else:
    print(year," is not a leap year")
```
