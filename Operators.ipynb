Introduction to Operators:-
1. Arithmatic Operation

```python
# Arithmatic operations
a=10
b=3
c = a + b  # Addition
d = a - b  # Subtraction
e = a * b  # Multiplication
f = a / b  # Division
g = a % b  # Modulus
h = a ** b  # Exponentiation
i=a//b # Floor division: remove the decimal value
print(c, d, e, f, g, h, i)
```

```python
## Comparision Operators
x = 10
y = 10
x == y  # Equal to
```

```python
## Not equal to
x=10
y=20
x != y  # Not equal to
```

```python
## greater than
x=10
y=20
x > y  # greater than
```

```python
## less than
x=10
y=20
x < y  # less than
```

```python
## greater than equal to
x=30
y=20
x >= y  # greater than equal to
```

```python
## less than equal to
x=30
y=20
x <= y  # less than equal to
```

```python
## logical operators ( And , or , not)
# And operator returns True if both operands are True, otherwise it returns False.
# or operator returns True if at least one of the operands is True, otherwise it returns False.
# not operator returns True if the operand is False, and returns False if the operand is True.
x = True
y = False
z = x and y  # Logical AND
print(z)  # Output: False

z = x or y  # Logical OR
print(z)  # Output: True

z = not x  # Logical NOT
print(z)  # Output: False
```
